## WEB 1

```python
from bs4 import BeautifulSoup 
import requests 
import pandas as pd
```

```python
url = "https://www.autofun.co.id/bengkel-motor/cari-surabaya-dan-dengan-review"
req = requests.get(url)
print(req)
```

```python
soup = BeautifulSoup(req.text, "html.parser")  
print(soup.link)
```

```python
nama = []
alamat = []
nomer = []
rating = []
review = []
komentar = []

for i in soup.find_all('a', class_ = 'station-item wa-g-unit'):
    nyaringalamat = i.find('p').text
    if 'Kota SBY' in nyaringalamat:
        
        
        nyaringnama = i.find('img').get('alt')
        nyaringrating = i.find('div', class_ = 'rate')

        
        if nyaringrating:
            nyaringrating2 = nyaringrating.text.split()
            nyaringrating3 = float(nyaringrating2[0].replace(',', '.'))
            nyaringreview = int(nyaringrating2[1].replace('(', '').replace(')', ''))
            
        else:
            nyaringrating3 = 0.0
            nyaringreview = 0

        url2 = requests.get('https://www.autofun.co.id' + i.get('href'))
        nyarilink = BeautifulSoup(url2.text, 'html.parser')
        
        nyaringnomor = nyarilink.find('div', class_ = 'phone info-item')
        if nyaringnomor:
            nyaringnomor = nyaringnomor.text 
        else:
            nyaringnomor = '-'
        
        
        for j in nyarilink.find_all('p', class_ = 'text wa-ellipsis-3'):
            nyaringkomentar = j.text 
            
        nama.append(nyaringnama)
        alamat.append(nyaringalamat)
        review.append(nyaringreview)
        rating.append(nyaringrating3)
        nomer.append(nyaringnomor)
        komentar.append(nyaringkomentar)
                
        
```

```python
df1 = pd.DataFrame({'Nama': nama, 'Alamat' : alamat, 'Review' : review, 'Nomer' : nomer, 'Rating' : rating, 'Komentar' : komentar })
df1
```

### STATISTIKA DESKRIPTIF

```python
df1.describe()
```

## WEB 2

```python
from bs4 import BeautifulSoup 
import requests 
import pandas as pd
```

```python
url = "https://jejakpiknik.com/bengkel-motor-surabaya/"
req = requests.get(url)
print(req)
```

```python
soup = BeautifulSoup(req.text, "html.parser")  
print(soup.link)
```

```python
nama = []
alamat = []
nomer = []
komentar = []

nemu = soup.find('div', class_ = 'entry-content')

for i in nemu.find_all('h3'):
    nyaringnama = i.text
    nyaringalamat = i.find_next('p').text.split('\n')[0][8:]
    nyaringnomer = i.find_next('p').text.split('\n')
    
    if len(nyaringnomer) == 4:
        nyaringnomer = nyaringnomer[3]
    else:
        nyaringnomer = '-'
        
    
    nyaringkomentar = i.find_next('p').find_next('p').text
    
    nama.append(nyaringnama)
    alamat.append(nyaringalamat)
    nomer.append(nyaringnomer)
    komentar.append(nyaringkomentar)   
    
```

```python
df2 = pd.DataFrame({'Nama': nama, 'Alamat' : alamat, 'Nomer' : nomer, 'Komentar' : komentar })
df2
```
